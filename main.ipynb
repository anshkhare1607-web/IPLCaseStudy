```python
# Task 0: Data Loading and Preprocessing
import numpy as np
```

```python
deliveries_headers = np.genfromtxt('deliveries.csv', delimiter=',', dtype=str, max_rows=1)
print("Deliveries Headers:\n", deliveries_headers)

matches_headers = np.genfromtxt('matches.csv', delimiter=',', dtype=str, max_rows=1)
print("\nMatches Headers:\n", matches_headers)
```

```python
deliveries_data = np.genfromtxt(
    'deliveries.csv', 
    delimiter=',', 
    dtype=str, 
    skip_header=1, 
    missing_values=['', 'NA'], 
    filling_values='0'
)

matches_data = np.genfromtxt(
    'matches.csv', 
    delimiter='|', 
    dtype=str, 
    skip_header=1, 
    missing_values=['', 'NA'], 
    filling_values='Unknown',
    invalid_raise=False 
)
```

```python
print(matches_data)
print()
print(deliveries_data)
```

```python
match_ids = deliveries_data[:, 0].astype(int)
batting_team = deliveries_data[:, 2]
over = deliveries_data[:, 4].astype(int)
batter = deliveries_data[:, 6]
bowler = deliveries_data[:, 7]
batsman_runs = deliveries_data[:, 9].astype(int)
```

```python
print("\n--- Extracted Arrays  ---")
print("match_ids:", match_ids[:50])
print("batting_team:", batting_team[:50])
print("batter:", batter[:50])
print("bowler:", bowler[:50])
print("batsman_runs:", batsman_runs[:50])
print("over:", over[:50])
```
