```python
# Task 0: Data Loading and Preprocessing
import numpy as np
```

```python
deliveries_headers = np.genfromtxt('deliveries.csv', delimiter=',', dtype=str, max_rows=1)
print("Deliveries Headers:\n", deliveries_headers)

matches_headers = np.genfromtxt('matches.csv', delimiter=',', dtype=str, max_rows=1)
print("\nMatches Headers:\n", matches_headers)
```

```python
deliveries_data = np.genfromtxt(
    'deliveries.csv', 
    delimiter=',', 
    dtype=str, 
    skip_header=1, 
    missing_values=['', 'NA'], 
    filling_values='0'
)

matches_data = np.genfromtxt(
    'matches.csv', 
    delimiter='|', 
    dtype=str, 
    skip_header=1, 
    missing_values=['', 'NA'], 
    filling_values='Unknown',
    invalid_raise=False 
)
```

```python
print(matches_data)
print()
print(deliveries_data)
```

```python
match_ids = deliveries_data[:, 0].astype(int)
batting_team = deliveries_data[:, 2]
over = deliveries_data[:, 4].astype(int)
batter = deliveries_data[:, 6]
bowler = deliveries_data[:, 7]
batsman_runs = deliveries_data[:, 9].astype(int)
```

```python
print("\n--- Extracted Arrays  ---")
print("match_ids:", match_ids[:50])
print("batting_team:", batting_team[:50])
print("batter:", batter[:50])
print("bowler:", bowler[:50])
print("batsman_runs:", batsman_runs[:50])
print("over:", over[:50])
```

```python
# Task 1: Total Runs per Match

#unique match IDs from the dataset
unique_match_ids = np.unique(match_ids)

runs_per_match = []

# 2. Get the list of unique match IDs
unique_match_ids = np.unique(match_ids)

# 3. Loop through every match one by one
for m in unique_match_ids:
    # Create a mask: True for the rows belonging to match 'm', False otherwise
    mask = (match_ids == m)
    
    # Filter: Use the mask to pick only the runs for this specific match
    match_runs_only = batsman_runs[mask]
    
    # Aggregate: Sum all those filtered runs
    total_runs = np.sum(match_runs_only)
    
    # Store the match ID and the total as a tuple in our list
    runs_per_match.append((m, total_runs))


# 3. Printing the output in the specified format [(match_id, total_runs), ...]
print("Required Output Format:")
print(runs_per_match[:5], "...") # Showing first 5 matches

# 4. Printing a neat summary table for the first 10 matches
print(f"\n{'Match ID':<10} | {'Total Runs':<10}")
print("-" * 25)
for match_id, total_runs in runs_per_match[:10]:
    print(f"{int(match_id):<10} | {int(total_runs):<10}")
```

```python
# Task 2: Top 5 Batters

# 1. unique batters list 
unique_batters = np.unique(batter)

# 2. total runs for each batter
batter_runs_list = []

for b in unique_batters:
    # Boolean Masking: 
    mask = (batter == b)
    # Aggregation: 
    total = np.sum(batsman_runs[mask])
    batter_runs_list.append(total)

# List 
batter_runs_array = np.array(batter_runs_list)

# 3. Ranking 
sorted_indices = np.argsort(batter_runs_array)

# 4. Top 5 
top_5_indices = sorted_indices[-5:][::-1]

# 5. Final list of tuples banate hain [(player_name, total_runs), ...]
top_5_batters = [(unique_batters[i], batter_runs_array[i]) for i in top_5_indices]

# --- Neat Output ---
print("Required Output Format:")
print(top_5_batters)

print(f"\n{'Player Name':<20} | {'Total Runs':<10}")
print("-" * 35)
for player, runs in top_5_batters:
    print(f"{player:<20} | {int(runs):<10}")
```
